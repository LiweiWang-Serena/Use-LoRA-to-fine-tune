```python
from unsloth import FastLanguageModel
from datasets import load_dataset
from trl import SFTTrainer
from transformers import TrainingArguments
import pandas as pd, re, gc, torch

model_name = "unsloth/Qwen2.5-7B-Instruct-bnb-4bit"
fine_tuning_file = "fine_tuning_data.jsonl"

test_cases = {
    "Severe dengue": "1D22",
    "Acute appendicitis": "DB10.0",
    "Chronic kidney disease": "GB61",
}

dataset_size = len(load_dataset("json", data_files=fine_tuning_file, split="train"))
print("Number of training examples:", dataset_size)
```

```python
def fine_tuning(input_model, tokenizer, batch_size, lr, steps, rank):
    # Add LoRA adapters to make it trainable
    model = FastLanguageModel.get_peft_model(input_model, r=rank,
           target_modules=["q_proj", "k_proj", "v_proj", "o_proj", "gate_proj", "up_proj", "down_proj"])

    # Load JSONL file
    dataset = load_dataset("json", data_files=fine_tuning_file, split="train")
    dataset = dataset.map(lambda x: {"text": tokenizer.apply_chat_template(x["messages"], tokenize=False)})

    # Train the model
    trainer = SFTTrainer(
        model=model,
        tokenizer=tokenizer,
        train_dataset=dataset,
        dataset_text_field="text",
        max_seq_length=2048,
        args=TrainingArguments(
            per_device_train_batch_size=batch_size,
            max_steps=steps,
            learning_rate=lr,
            logging_steps=10,
            output_dir="outputs",
            report_to="none",
        ),
    )
    trainer.train()
    final_loss = [x["loss"] for x in trainer.state.log_history if "loss" in x][-1]
    return model, final_loss


def generate_answer(gen_model, tokenizer, user_query):
    FastLanguageModel.for_inference(gen_model)
    messages = [{"role": "user", "content": user_query}]
    text = tokenizer.apply_chat_template(messages, tokenize=False, add_generation_prompt=True)
    inputs = tokenizer(text, return_tensors="pt").to("cuda")
    outputs = gen_model.generate(**inputs, max_new_tokens=20, do_sample=False)
    # Keep only the newly generated tokens (drop the prompt)
    return tokenizer.decode(outputs[0][inputs["input_ids"].shape[1]:], skip_special_tokens=True).strip()


CODE_RE = re.compile(r"\b[0-9A-Z][A-Z][0-9][0-9A-Z](?:\.[0-9A-Z]{1,2})?\b")

def test_model(gen_model, tokenizer):
    predictions, correct = {}, 0
    for disease, expected in test_cases.items():
        answer = generate_answer(gen_model, tokenizer, f"what is the ICD11 code of {disease}?")
        match = CODE_RE.search(answer.upper())
        predicted = match.group(0) if match else ""
        ok = predicted == expected
        correct += ok
        print(f"{disease}: {answer}  ->  {'correct' if ok else 'wrong'} (expected {expected})")
        predictions[disease] = predicted
    return predictions, correct
```

```python
model, tokenizer = FastLanguageModel.from_pretrained(model_name=model_name)
test_model(model, tokenizer)
del model, tokenizer
gc.collect(); torch.cuda.empty_cache()
```

```python
settings = [
    # (batch_size, lr, steps, rank)
    (4, 4e-4, 100, 32),                         # baseline (sample code values)
    (2, 4e-4, 100, 32), (8, 4e-4, 100, 32),     # batch size
    (4, 1e-4, 100, 32), (4, 8e-4, 100, 32),     # learning rate
    (4, 4e-4, 50, 32),  (4, 4e-4, 200, 32),     # steps
    (4, 4e-4, 100, 8),  (4, 4e-4, 100, 64),     # rank
]

records = []
for i, (batch_size, lr, steps, rank) in enumerate(settings, 1):
    epochs = batch_size * steps / dataset_size
    print(f"\n===== Setting {i}: batch_size={batch_size}, lr={lr}, steps={steps}, rank={rank} (epochs = {epochs:.2f}) =====")
    model, tokenizer = FastLanguageModel.from_pretrained(model_name=model_name)
    fine_tuned_model, final_loss = fine_tuning(model, tokenizer, batch_size, lr, steps, rank)
    predictions, correct = test_model(fine_tuned_model, tokenizer)
    records.append({"setting": i, "batch_size": batch_size, "lr": lr, "steps": steps, "rank": rank,
                    "final_loss": round(final_loss, 4), **predictions, "correct (of 3)": correct})
    pd.DataFrame(records).to_csv("results.csv", index=False)   # save progress after every run
    del model, fine_tuned_model, tokenizer
    gc.collect(); torch.cuda.empty_cache()
```

```python
results_df = pd.DataFrame(records)
display(results_df)

best = results_df.sort_values(["correct (of 3)", "final_loss"], ascending=[False, True]).iloc[0]
print("Best setting:", best[["batch_size", "lr", "steps", "rank"]].to_dict())
print("Correct:", best["correct (of 3)"], "/ 3 | Final training loss:", best["final_loss"])
```

## Report

**Best hyper-parameters: batch size = 4, learning rate = 4e-4, steps = 200, rank = 32**
This setting answered all three test diseases correctly (3/3) and had the lowest final training loss (0.1583).

**Observations**
- Before fine-tuning, the base model did not return any ICD-11 code. After fine-tuning, all settings learned the answer format "ICD11:XXXX".
- The number of epochs (batch size × steps ÷ 200) was the most important factor. Only the two settings trained for 4 epochs (setting 3: batch size 8, and setting 7: 200 steps) gave correct codes. All settings with 1–2 epochs scored 0/3.
- Final training loss matched the accuracy: the two lowest losses (0.1583 and 0.1747) were the only settings with correct answers.
- With the same 4 epochs, 200 steps with batch size 4 (3/3) was better than 100 steps with batch size 8 (2/3), suggesting that more parameter updates help memorize the codes.
- Rank was not the bottleneck: rank 8, 32, and 64 produced the same predictions and similar losses (about 0.26), so even rank 8 has enough capacity for 200 codes.
- A small learning rate (1e-4) learned too slowly (loss 0.3498), while a larger learning rate (8e-4) did not help when training was limited to 2 epochs.
- Wrong answers were often close to the correct code (e.g., GB60 vs. GB61), showing that the model first learns the format, then the chapter prefix, and finally the exact code.

**Limitations**
The test diseases are included in the training data, so this experiment measures memorization rather than generalization to unseen diseases. With only three test cases and one run per setting, the results may vary with different random seeds.
